## Baselines

```python
from sklearn.dummy import DummyClassifier
from sklearn.model_selection import cross_validate

metriques = ["f1_macro", "recall", "precision", "accuracy", "roc_auc"]
resultats_dummy = {}
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
for strat in ["most_frequent", "stratified"]:
    dummy = DummyClassifier(strategy=strat, random_state=42)
    resultats_dummy[strat] = cross_validate(dummy, X_train, y_train, cv=cv, scoring=metriques)
    print(f"\nBaseline « {strat} »")
    for m in metriques:
        v = resultats_dummy[strat][f"test_{m}"]
        print(f"  {m:10s} : {v.mean():.3f} ± {v.std():.3f}")
```
